```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
```

```python
# SECCION DE COMPARACION DE CSV
sub_anterior = pd.read_csv("submissions/mas_probable_cobro_ROI.csv")
sub_nuevo = pd.read_csv("submissions/mas_probable_cobro_ROIMLP.csv")

# 1. Asegurar el mismo orden de IDs
sub_anterior = sub_anterior.sort_values(by="ID_Solicitud").reset_index(drop=True)
sub_nuevo = sub_nuevo.sort_values(by="ID_Solicitud").reset_index(drop=True)

# 2. Conteo de coincidencias y discrepancias en las decisiones
coincidencias = (sub_anterior["Decision_Aprobacion"] == sub_nuevo["Decision_Aprobacion"]).sum()
discrepancias = (sub_anterior["Decision_Aprobacion"] != sub_nuevo["Decision_Aprobacion"]).sum()
total_filas = len(sub_nuevo)

print(f"Total filas: {total_filas}")
print(f"Decisiones idénticas: {coincidencias} ({coincidencias / total_filas * 100:.2f}%)")
print(f"Decisiones diferentes: {discrepancias} ({discrepancias / total_filas * 100:.2f}%)")

# 3. Ver cuáles cambiaron de estado
cambios = sub_anterior[["ID_Solicitud"]].copy()
cambios["Decision_Anterior"] = sub_anterior["Decision_Aprobacion"]
cambios["Decision_Nueva"] = sub_nuevo["Decision_Aprobacion"]

filtro_cambios = cambios[cambios["Decision_Anterior"] != cambios["Decision_Nueva"]]
print("\nPrimeras solicitudes que cambiaron de decisión:")
display(filtro_cambios.head(10))

# 4. Detalle de cambios de aprobación/rechazo
de_rechazado_a_aprobado = ((cambios["Decision_Anterior"] == 0) & (cambios["Decision_Nueva"] == 1)).sum()
de_aprobado_a_rechazado = ((cambios["Decision_Anterior"] == 1) & (cambios["Decision_Nueva"] == 0)).sum()

print(f"\nPréstamos que antes se rechazaban y ahora se aprueban: {de_rechazado_a_aprobado}")
print(f"Préstamos que antes se aprobaban y ahora se rechazan: {de_aprobado_a_rechazado}")
```
